```python
import nltk
from nltk.chat.util import Chat, reflections

relfections={
    "I am": "You are",
    "I was": "You were",
    "I": "You",
    "I'm" : "You are",
    "I've": "You have",
    "I'll": "You will",
    "My": "Your",
    "You are": "I am",
    "You were": "I was",
    "You": "I",
    "You've": "I have",
    "you'll": "I will",
    "your": "My",
    "Yours": "Mine",
    "you": "me",
    "Me": "You"
}

pairs=[
    [
        r"My name is (.*)",
        ["Hello %1, How are you today?",]
    ],
    [
        r"Hi/Hey/Hello",
        ["Hello", "Hey there",]
    ],
    [
        r"What is your name?",
        ["Hello my name is Javis"]
    ]
]

def chat():
  print("Hi there, how may I assist you?: ")
  chat=Chat(pairs,reflections)
  chat.converse()
  return

chat()
```
